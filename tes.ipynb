```python

import pandas as pd
import numpy as np

# ==== 1. LOAD DATA ====
# If your file is CSV, for example:
# df = pd.read_csv("xauusd_m1.csv")
# It must contain: open, high, low, close, tick_volume, spread, real_volume

df = pd.read_csv("data.csv")  # change to your file name

# ==== 2. PARAMETERS ====
initial_capital = 1000.0          # USD
pip_size = 0.1                    # 1 pip = 0.1 if XAUUSD has 1 decimal; adjust to your broker
stop_pips = 12.5                  # 10–15 pips -> choose midpoint
risk_per_trade = 0.01             # 1% of capital per trade

short_ema_period = 9
long_ema_period = 21
rsi_period = 14
bb_period = 20
bb_std = 2

# ==== 3. INDICATORS ====
# EMAs
df["ema_fast"] = df["close"].ewm(span=short_ema_period, adjust=False).mean()
df["ema_slow"] = df["close"].ewm(span=long_ema_period, adjust=False).mean()

# RSI (classic Wilder-style using rolling mean; for scalping you might later switch to EMA)
delta = df["close"].diff()
up = delta.clip(lower=0)
down = -delta.clip(upper=0)
roll_up = up.rolling(rsi_period).mean()
roll_down = down.rolling(rsi_period).mean()
rs = roll_up / roll_down
df["rsi"] = 100.0 - (100.0 / (1.0 + rs))

# Bollinger Bands
ma = df["close"].rolling(window=bb_period).mean()
std = df["close"].rolling(window=bb_period).std()
df["bb_mid"] = ma
df["bb_upper"] = ma + bb_std * std
df["bb_lower"] = ma - bb_std * std


# ==== 4. BACKTEST LOOP ====

position = 0           # 0 = flat, 1 = long, -1 = short
entry_price = 0.0
entry_index = None
capital = initial_capital
equity_curve = []

trade_log = []

for i in range(1, len(df)):
    row_prev = df.iloc[i - 1]
    row = df.iloc[i]

    # Skip until indicators valid
    if (
        np.isnan(row_prev["ema_fast"]) or np.isnan(row_prev["ema_slow"])
        or np.isnan(row_prev["rsi"]) or np.isnan(row_prev["bb_lower"])
        or np.isnan(row_prev["bb_upper"])
    ):
        equity_curve.append(capital)
        continue

    # ===== MANAGE OPEN POSITION =====
    if position != 0:
        if position == 1:
            # Long: stop below entry, TP at upper band, RSI>=70
            stop_loss = entry_price - stop_pips * pip_size
            take_profit_band = row["bb_upper"]
            exit_rsi = row["rsi"] >= 70

            exit_reason = None
            exit_price = None

            # Check SL first
            if row["low"] <= stop_loss:
                exit_price = stop_loss
                exit_reason = "SL"
            # Then TP / RSI
            elif (row["high"] >= take_profit_band) or exit_rsi:
                # Assume you get filled at band if touched, else close price
                if row["high"] >= take_profit_band:
                    exit_price = take_profit_band
                    exit_reason = "TP_band"
                else:
                    exit_price = row["close"]
                    exit_reason = "RSI_exit"

            if exit_price is not None:
                # position size in units of XAU based on 1% risk and stop distance
                risk_amount = capital * risk_per_trade
                risk_per_unit = abs(entry_price - stop_loss)
                units = risk_amount / risk_per_unit if risk_per_unit > 0 else 0

                pnl = (exit_price - entry_price) * units
                capital += pnl

                trade_log.append({
                    "side": "long",
                    "entry_index": entry_index,
                    "exit_index": i,
                    "entry_price": entry_price,
                    "exit_price": exit_price,
                    "units": units,
                    "pnl": pnl,
                    "capital_after": capital,
                    "exit_reason": exit_reason,
                })

                position = 0
                entry_price = 0.0
                entry_index = None

        elif position == -1:
            # Short: stop above entry, TP at lower band, RSI<=30
            stop_loss = entry_price + stop_pips * pip_size
            take_profit_band = row["bb_lower"]
            exit_rsi = row["rsi"] <= 30

            exit_reason = None
            exit_price = None

            # Check SL first
            if row["high"] >= stop_loss:
                exit_price = stop_loss
                exit_reason = "SL"
            # Then TP / RSI
            elif (row["low"] <= take_profit_band) or exit_rsi:
                if row["low"] <= take_profit_band:
                    exit_price = take_profit_band
                    exit_reason = "TP_band"
                else:
                    exit_price = row["close"]
                    exit_reason = "RSI_exit"

            if exit_price is not None:
                risk_amount = capital * risk_per_trade
                risk_per_unit = abs(stop_loss - entry_price)
                units = risk_amount / risk_per_unit if risk_per_unit > 0 else 0

                pnl = (entry_price - exit_price) * units
                capital += pnl

                trade_log.append({
                    "side": "short",
                    "entry_index": entry_index,
                    "exit_index": i,
                    "entry_price": entry_price,
                    "exit_price": exit_price,
                    "units": units,
                    "pnl": pnl,
                    "capital_after": capital,
                    "exit_reason": exit_reason,
                })

                position = 0
                entry_price = 0.0
                entry_index = None

    # ===== LOOK FOR NEW ENTRIES IF FLAT =====
    if position == 0:
        crossed_up = (row_prev["ema_fast"] <= row_prev["ema_slow"]) and (row["ema_fast"] > row["ema_slow"])
        crossed_down = (row_prev["ema_fast"] >= row_prev["ema_slow"]) and (row["ema_fast"] < row["ema_slow"])

        price_near_lower = row["close"] <= row["bb_lower"] * 1.005   # within ~0.5%
        price_near_upper = row["close"] >= row["bb_upper"] * 0.995   # within ~0.5%

        rsi_ok = (row["rsi"] > 30) and (row["rsi"] < 70)

        # Long entry
        if crossed_up and rsi_ok and price_near_lower:
            position = 1
            entry_price = row["close"]
            entry_index = i

        # Short entry
        elif crossed_down and rsi_ok and price_near_upper:
            position = -1
            entry_price = row["close"]
            entry_index = i

    equity_curve.append(capital)

# Convert trade log to DataFrame for analysis
trades = pd.DataFrame(trade_log)

print("Initial capital:", initial_capital)
print("Final capital:", capital)
print("Number of trades:", len(trades))
print("Win rate:", (trades["pnl"] > 0).mean() if len(trades) > 0 else 0)

# Optional: save trades
trades.to_csv("xauusd_ma_rsi_bb_trades.csv", index=False)
```

```python
import pandas as pd
import numpy as np
import os
from decimal import Decimal, getcontext

# High precision
getcontext().prec = 10

# ------------- CONFIG (XAUUSD SPOT) -------------
csv_path = "5min_data.csv"          # your XAUUSD M5 OHLC data
initial_capital = Decimal("5000.0")

# One "unit" = 0.01 lot if your broker uses 1.0 lot = 100 oz
lot_size = Decimal("0.15")          # 0.10 lot per trade (example)
contract_oz_per_lot = Decimal("100")  # standard XAUUSD: 1 lot = 100 oz

# Risk/Reward in dollars per oz
tp_points = Decimal("1.5")          # $1.5 move in gold price
sl_points = Decimal("0.5")          # $0.5 move in gold price

charges_roundtrip = Decimal("0.0")  # commissions + fees per trade
entry_buffer = Decimal("0.00")      # small buffer above/below previous bar
# -----------------------------------------------

if not os.path.exists(csv_path):
    raise FileNotFoundError(f"CSV not found: {csv_path}")

df = pd.read_csv(csv_path, dtype=str)
df["timestamp"] = pd.to_datetime(df["datetime"])

for c in ["open", "high", "low", "close"]:
    df[c] = df[c].apply(lambda x: Decimal(str(x)))

df = df.sort_values("timestamp").reset_index(drop=True)

capital = initial_capital
equity = [capital]
position = None
trades = []

for i in range(1, len(df)):
    prev = df.loc[i - 1]
    curr = df.loc[i]
    time = curr["timestamp"].time()

    # Adjust to your trading hours if needed (XAUUSD is almost 24h)
    # Example: trade only London+NY (12:00–21:00 IST approx)
    if not (pd.to_datetime("12:00").time() <= time <= pd.to_datetime("21:00").time()):
        equity.append(capital)
        continue

    # Manage open position
    if position is not None:
        side = position["side"]
        entry_price = position["entry_price"]

        tp = entry_price + tp_points if side == "long" else entry_price - tp_points
        sl = entry_price - sl_points if side == "long" else entry_price + sl_points

        hit_target = (side == "long" and curr["high"] >= tp) or (side == "short" and curr["low"] <= tp)
        hit_stop = (side == "long" and curr["low"] <= sl) or (side == "short" and curr["high"] >= sl)

        if hit_stop and hit_target:
            exit_price, reason = sl, "stop_loss (both hit)"
        elif hit_target:
            exit_price, reason = tp, "target"
        elif hit_stop:
            exit_price, reason = sl, "stop_loss"
        else:
            equity.append(capital)
            continue

        # P&L in dollars
        price_move = (exit_price - entry_price) if side == "long" else (entry_price - exit_price)
        # value per 1.0 price move = lot_size * contract_oz_per_lot
        pnl = price_move * lot_size * contract_oz_per_lot
        pnl -= charges_roundtrip
        capital += pnl

        trades.append({
            "entry_time": position["entry_time"],
            "exit_time": curr["timestamp"],
            "side": side,
            "entry_price": float(entry_price),
            "exit_price": float(exit_price),
            "pnl": float(pnl),
            "exit_reason": reason,
            "capital_after": float(capital)
        })
        position = None
        equity.append(capital)
        continue

    # Entry logic – breakout of previous candle high/low with buffer
    if curr["close"] > prev["high"] + entry_buffer:
        position = {
            "side": "long",
            "entry_price": curr["open"],
            "entry_time": curr["timestamp"]
        }
    elif curr["close"] < prev["low"] - entry_buffer:
        position = {
            "side": "short",
            "entry_price": curr["open"],
            "entry_time": curr["timestamp"]
        }

    equity.append(capital)

# End-of-data close
if position is not None:
    last_price = df.iloc[-1]["close"]
    side = position["side"]
    entry_price = position["entry_price"]
    price_move = (last_price - entry_price) if side == "long" else (entry_price - last_price)
    pnl = price_move * lot_size * contract_oz_per_lot - charges_roundtrip
    capital += pnl
    trades.append({
        "entry_time": position["entry_time"],
        "exit_time": df.iloc[-1]["timestamp"],
        "side": side,
        "entry_price": float(entry_price),
        "exit_price": float(last_price),
        "pnl": float(pnl),
        "exit_reason": "eod_close",
        "capital_after": float(capital)
    })

# ------------- SUMMARY -------------
trades_df = pd.DataFrame(trades)
if not trades_df.empty:
    total_pnl = trades_df["pnl"].sum()
    win_rate = 100 * (trades_df["pnl"] > 0).sum() / len(trades_df)
    avg_pnl = trades_df["pnl"].mean()
    dd = pd.Series(equity) - pd.Series(equity).cummax()
    max_dd = dd.min()

    print(f"Initial capital: ${initial_capital}")
    print(f"Lot size: {lot_size} | Contract: {contract_oz_per_lot} oz/lot")
    print(f"Final capital:   ${capital}")
    print(f"Net P&L:         ${total_pnl:.2f}")
    print(f"Trades:          {len(trades_df)} | Win rate: {win_rate:.1f}%")
    print(f"Average P&L:     ${avg_pnl:.2f}")
    print(f"Max Drawdown:    ${max_dd}\n")
    print("Exit Reasons:\n", trades_df["exit_reason"].value_counts())
    print("\nSample Trades:\n", trades_df.head(10))
else:
    print("No trades executed — check breakout conditions or data / hours.")
```
